print FIZZ if a number is divisible by only 3,print BUZZ if a num is divisible #by only 5 print FIZZBUZZ if it is divisible by both 3 and 5

```python
n=int(input("enter a num:"))
if(n%3==0 and n%5!=0):
  print("FIZZ")
elif(n%5==0 and n%3!=0):
  print("BUZZ")

elif(n%3==0 and n%5==0):
    print("FIZZBUZZ")
```
